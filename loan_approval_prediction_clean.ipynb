# Loan Approval Prediction

This project uses applicant information to predict whether a loan application is approved or rejected using Logistic Regression.

**Technologies:** Python, Pandas, Scikit-learn


## 1. Import Libraries


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
```

## 2. Load the Dataset

The dataset contains applicant information and the loan approval status.


```python
df = pd.read_csv("loan_data.csv")
df.head()
```

## 3. Check the Data


```python
print("Shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
```

## 4. Handle Missing Values

Missing values are replaced with 0 for this small practice dataset.


```python
df = df.fillna(0)
```

## 5. Convert Categorical Values to Numbers

The machine learning model works with numerical values, so categorical columns are encoded using simple mappings.


```python
df['Gender'] = df['Gender'].map({'Male': 1, 'Female': 0})
df['Married'] = df['Married'].map({'Yes': 1, 'No': 0})
df['Education'] = df['Education'].map({'Graduate': 1, 'Not Graduate': 0})
df['Loan_Status'] = df['Loan_Status'].map({'Y': 1, 'N': 0})

df.head()
```

## 6. Separate Features and Target

`X` contains the applicant information used for prediction, while `y` contains the loan approval status.


```python
X = df.drop('Loan_Status', axis=1)
y = df['Loan_Status']
```

## 7. Split the Data

The data is divided into training and testing sets. A fixed random state makes the split reproducible.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## 8. Train the Logistic Regression Model


```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

## 9. Make Predictions


```python
y_pred = model.predict(X_test)
print("Predicted values:", y_pred)
```

## 10. Evaluate the Model

Accuracy shows the proportion of test predictions that were correct. The confusion matrix shows the prediction counts by class.


```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

## Note

This is a small practice dataset, so the evaluation result should not be treated as a reliable measure of real-world loan approval performance. A larger and more representative dataset would be needed for a production-level model.
